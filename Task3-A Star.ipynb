```python
import heapq

def h(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def astar(start, goal, grid):
    q = [(h(start, goal), 0, start)]
    parent = {start: None}
    cost = {start: 0}
    path = []

    while q:
        f, g, p = heapq.heappop(q)

        if p == goal:
            while p is not None:
                path.append(p)
                p = parent[p]
            return path[::-1], g

        for dx, dy in [(1, 0), (-1, 0), (0, 1), (0, -1)]:
            x = p[0] + dx
            y = p[1] + dy

            if 0 <= x < len(grid) and 0 <= y < len(grid[0]):
                if grid[x][y] == 1:
                    continue

                ng = g + 1

                if (x, y) not in cost or ng < cost[(x, y)]:
                    cost[(x, y)] = ng
                    parent[(x, y)] = p
                    heapq.heappush(
                        q,
                        (ng + h((x, y), goal), ng, (x, y))
                    )

    return None, None


grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]

start = (0, 0)
goal = (2, 2)

path, total_cost = astar(start, goal, grid)

print("Input Grid:")
for row in grid:
    print(row)

print("Start:", start)
print("Goal:", goal)
print("Optimal Path:", path)
print("Cost:", total_cost)
```
