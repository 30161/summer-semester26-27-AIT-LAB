```python
from collections import deque

def bfsofgraph(v, adj):
    bfs = []
    vis = [False] * v

    q = deque([0])
    vis[0] = True

    while q:
        node = q.popleft()
        bfs.append(node)

        for it in adj[node]:
            if not vis[it]:
                vis[it] = True
                q.append(it)

    return bfs

    adj = [
    [1, 2],
    [0, 3, 4],
    [0],
    [1],
    [1]
]

print(bfsofgraph(5, adj))

```

```python
def dfs(node, adj, vis, ans):
    vis[node] = True
    ans.append(node)

    for it in adj[node]:
        if not vis[it]:
            dfs(it, adj, vis, ans)


def dfsOfGraph(v, adj):
    vis = [False] * v
    ans = []

    dfs(0, adj, vis, ans)

    return ans

adj = [
    [1, 2],
    [0, 3, 4],
    [0],
    [1],
    [1]
]

print(dfsOfGraph(5, adj))
```
