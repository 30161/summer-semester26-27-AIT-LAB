```python
R1 = float(input("Enter value of Path 1: "))
R2 = float(input("Enter value of Path 2: "))

total = R1 + R2

P1 = R1 / total
P2 = R2 / total

print("\n--- ACO Result ---")

print("Total =", total)

print("P1 =", P1, "or", P1 * 100, "%")
print("P2 =", P2, "or", P2 * 100, "%")

if P1 > P2:
    print("Most ants followed Path 1.")
    print("Path 1 has the higher probability.")
elif P2 > P1:
    print("Most ants followed Path 2.")
    print("Path 2 has the higher probability.")
else:
    print("Both paths have equal probability.")

print("P1 + P2 =", P1 + P2)

import matplotlib.pyplot as plt

paths = ["Path 1", "Path 2"]
probabilities = [P1 * 100, P2 * 100]

plt.bar(paths, probabilities)

plt.title("Ant Colony Optimization Probability Graph")
plt.xlabel("Paths")
plt.ylabel("Probability (%)")

plt.ylim(0, 100)

for i, value in enumerate(probabilities):
    plt.text(i, value + 2, f"{value:.2f}%", ha='center')

plt.show()
```
